```python
import torch
import torch.nn as nn
```

```python
class Perceptron(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.linear = nn.Linear(input_size, 1)

    def forward(self, x):
        z = self.linear(x)
        return (z >= 0).float()

# AND gate dataset
X = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y = torch.tensor([[0.], [0.], [0.], [1.]])

model = Perceptron(input_size=2)

# Perceptron learning algorithm
lr = 0.1

with torch.no_grad():
    for epoch in range(100):
        for xi, yi in zip(X, y):
            z = model.linear(xi)
            pred = (z >= 0).float()
            error = yi - pred

            model.linear.weight += lr * error * xi.unsqueeze(0)
            model.linear.bias += lr * error.squeeze()

print("Predictions:")
print(model(X))
```
